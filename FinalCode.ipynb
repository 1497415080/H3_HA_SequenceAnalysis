```python
import os
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter

sns.set(style="whitegrid", context="talk")


# ============================================================
# 1. 输入和输出路径
# ============================================================

FASTA_FILE = "HA_20260920-Nvirus.year.1968-2026.fasta"

OUTDIR = "year_consensus"
FIGDIR = os.path.join(OUTDIR, "figures")

os.makedirs(OUTDIR, exist_ok=True)
os.makedirs(FIGDIR, exist_ok=True)

LONG_FILE = os.path.join(OUTDIR, "yearly_consensus_long.tsv")
AA_MATRIX_FILE = os.path.join(OUTDIR, "yearly_consensus_AA_matrix.tsv")
PERCENT_MATRIX_FILE = os.path.join(OUTDIR, "yearly_consensus_percent_matrix.tsv")
COUNT_FILE = os.path.join(OUTDIR, "yearly_sequence_counts.tsv")
POSMAP_FILE = os.path.join(OUTDIR, "alignment_to_reference_position_map.tsv")


# ============================================================
# 2. 参数设置
# ============================================================

# 是否忽略 gap 计算 consensus
IGNORE_GAP_FOR_CONSENSUS = True

# 是否忽略 X / ? / . 这类未知氨基酸
IGNORE_UNKNOWN_FOR_CONSENSUS = False

UNKNOWN_AA = {"X", "?", "."}

# 如果同一年同一位点出现并列最高频，如何处理
# "alphabetical": 按字母顺序选第一个
# "reference": 优先选择参考序列该位点氨基酸，如果参考氨基酸也在并列最高频中
TIE_BREAK_MODE = "reference"

# 参考序列：默认使用 FASTA 中第一条序列
REFERENCE_MODE = "first"


# ============================================================
# 3. FASTA 读取函数
# ============================================================

def read_fasta(fasta_file):
    """
    读取 fasta 文件，返回 list of dict:
    [
        {"header": "...", "seq": "..."},
        ...
    ]
    """
    records = []
    header = None
    seq_lines = []

    with open(fasta_file, "r", encoding="utf-8") as f:
        for line in f:
            line = line.strip()

            if not line:
                continue

            if line.startswith(">"):
                if header is not None:
                    records.append({
                        "header": header,
                        "seq": "".join(seq_lines).upper()
                    })

                header = line[1:].strip()
                seq_lines = []
            else:
                seq_lines.append(line.strip())

        if header is not None:
            records.append({
                "header": header,
                "seq": "".join(seq_lines).upper()
            })

    return records


def parse_year_from_header(header):
    """
    从 header 中解析年份。

    适配你的格式：
    >YP_308839.1 | 2004
    >BFM37792.1 | 2023

    也兼容 header 中其他位置出现 4 位年份的情况。
    """
    # 优先解析竖线后面的年份
    m = re.search(r"\|\s*(\d{4})\s*$", header)
    if m:
        return int(m.group(1))

    # 如果没有竖线，则搜索任意 4 位年份
    m = re.search(r"(19\d{2}|20\d{2})", header)
    if m:
        return int(m.group(1))

    return None


def parse_accession_from_header(header):
    """
    从 header 中解析 accession。
    例如：
    YP_308839.1 | 2004 -> YP_308839.1
    """
    if "|" in header:
        return header.split("|")[0].strip()
    return header.split()[0].strip()


# ============================================================
# 4. 读取 FASTA 并整理为 DataFrame
# ============================================================

records = read_fasta(FASTA_FILE)

if len(records) == 0:
    raise ValueError(f"没有从 {FASTA_FILE} 中读取到任何序列。")

data = []

for rec in records:
    header = rec["header"]
    seq = rec["seq"]

    year = parse_year_from_header(header)
    accession = parse_accession_from_header(header)

    if year is None:
        print(f"警告：无法解析年份，跳过该序列：{header}")
        continue

    data.append({
        "accession": accession,
        "year": year,
        "header": header,
        "sequence": seq,
        "length": len(seq)
    })

seq_df = pd.DataFrame(data)

if seq_df.empty:
    raise ValueError("所有序列都无法解析年份，请检查 FASTA header 格式。")

seq_df = seq_df.sort_values(["year", "accession"]).reset_index(drop=True)

print("成功读取序列数:", len(seq_df))
print("年份范围:", seq_df["year"].min(), "-", seq_df["year"].max())
print("序列长度分布:")
display(seq_df["length"].value_counts().sort_index())


# ============================================================
# 5. 检查序列是否已经等长对齐
# ============================================================

lengths = seq_df["length"].unique()

if len(lengths) != 1:
    raise ValueError(
        "检测到序列长度不一致。\n"
        "当前脚本要求输入 FASTA 已经是多序列比对后的 aligned fasta，或者所有序列长度一致。\n"
        "请先用 MAFFT / MUSCLE / Clustal Omega 对序列进行比对。\n"
        f"检测到的长度有: {sorted(lengths)}"
    )

alignment_length = int(lengths[0])
print("alignment length:", alignment_length)


# ============================================================
# 6. 选择参考序列，并生成 alignment -> reference position map
# ============================================================

if REFERENCE_MODE == "first":
    ref_seq = seq_df.iloc[0]["sequence"]
    ref_header = seq_df.iloc[0]["header"]
else:
    raise ValueError("目前 REFERENCE_MODE 仅支持 'first'。")

pos_map_records = []
ref_pos = 0

for aln_pos, aa in enumerate(ref_seq, start=1):
    if aa != "-":
        ref_pos += 1
        reference_position = ref_pos
    else:
        reference_position = np.nan

    pos_map_records.append({
        "alignment_position": aln_pos,
        "reference_position": reference_position,
        "reference_AA": aa
    })

pos_map = pd.DataFrame(pos_map_records)

pos_map.to_csv(POSMAP_FILE, sep="\t", index=False)

print("参考序列:", ref_header)
print("参考序列非 gap 长度:", int(pos_map["reference_position"].notna().sum()))
print("pos map saved to:", POSMAP_FILE)


# ============================================================
# 7. 每年序列数量统计
# ============================================================

year_counts = (
    seq_df
    .groupby("year")
    .size()
    .reset_index(name="n_sequences")
    .sort_values("year")
)

year_counts.to_csv(COUNT_FILE, sep="\t", index=False)

print("year counts saved to:", COUNT_FILE)
display(year_counts.head())


# ============================================================
# 8. 计算每一年每个位点的 consensus
# ============================================================

def consensus_at_position(chars, ref_char=None):
    """
    输入某一年某个位点的所有字符，返回 consensus 信息。
    """
    original_n = len(chars)

    filtered = []

    for c in chars:
        c = c.upper()

        if IGNORE_GAP_FOR_CONSENSUS and c == "-":
            continue

        if IGNORE_UNKNOWN_FOR_CONSENSUS and c in UNKNOWN_AA:
            continue

        filtered.append(c)

    n_used = len(filtered)

    if n_used == 0:
        return {
            "consensus_AA": "-",
            "consensus_count": 0,
            "n_used": 0,
            "consensus_percent": np.nan,
            "residue_counts": ""
        }

    counter = Counter(filtered)
    max_count = max(counter.values())

    candidates = sorted([aa for aa, count in counter.items() if count == max_count])

    if TIE_BREAK_MODE == "reference" and ref_char is not None:
        if ref_char in candidates:
            consensus_aa = ref_char
        else:
            consensus_aa = candidates[0]
    else:
        consensus_aa = candidates[0]

    consensus_count = counter[consensus_aa]
    consensus_percent = consensus_count / n_used * 100

    residue_counts = ";".join(
        [f"{aa}:{counter[aa]}" for aa in sorted(counter.keys())]
    )

    return {
        "consensus_AA": consensus_aa,
        "consensus_count": consensus_count,
        "n_used": n_used,
        "consensus_percent": consensus_percent,
        "residue_counts": residue_counts
    }


long_records = []

years = sorted(seq_df["year"].unique())

for year in years:
    sub = seq_df[seq_df["year"] == year]
    seqs = sub["sequence"].tolist()
    n_sequences = len(seqs)

    for aln_pos in range(1, alignment_length + 1):
        idx = aln_pos - 1

        chars = [seq[idx] for seq in seqs]
        ref_char = ref_seq[idx]

        cinfo = consensus_at_position(chars, ref_char=ref_char)

        ref_position = pos_map.loc[
            pos_map["alignment_position"] == aln_pos,
            "reference_position"
        ].iloc[0]

        reference_AA = pos_map.loc[
            pos_map["alignment_position"] == aln_pos,
            "reference_AA"
        ].iloc[0]

        long_records.append({
            "year": year,
            "alignment_position": aln_pos,
            "reference_position": ref_position,
            "reference_AA": reference_AA,
            "n_sequences": n_sequences,
            "n_used_for_consensus": cinfo["n_used"],
            "consensus_AA": cinfo["consensus_AA"],
            "consensus_count": cinfo["consensus_count"],
            "consensus_percent": cinfo["consensus_percent"],
            "residue_counts": cinfo["residue_counts"]
        })

long_df = pd.DataFrame(long_records)

long_df.to_csv(LONG_FILE, sep="\t", index=False)

print("long consensus table saved to:", LONG_FILE)
display(long_df.head())


# ============================================================
# 9. 生成 consensus AA matrix 和 percent matrix
# ============================================================

# 列名使用 alignment_position
# 这样可以和 alignment_to_reference_position_map.tsv 对应
aa_mat = (
    long_df
    .pivot(index="year", columns="alignment_position", values="consensus_AA")
    .sort_index()
)

percent_mat = (
    long_df
    .pivot(index="year", columns="alignment_position", values="consensus_percent")
    .sort_index()
)

# 列名转成字符串，方便保存和后续读取
aa_mat.columns = aa_mat.columns.astype(str)
percent_mat.columns = percent_mat.columns.astype(str)

aa_mat.to_csv(AA_MATRIX_FILE, sep="\t")
percent_mat.to_csv(PERCENT_MATRIX_FILE, sep="\t")

print("AA matrix saved to:", AA_MATRIX_FILE)
print("percent matrix saved to:", PERCENT_MATRIX_FILE)

display(aa_mat.head())
display(percent_mat.head())


# ============================================================
# 10. 重新读取确认
# ============================================================

long_df = pd.read_csv(LONG_FILE, sep="\t")
aa_mat = pd.read_csv(AA_MATRIX_FILE, sep="\t", index_col=0)
percent_mat = pd.read_csv(PERCENT_MATRIX_FILE, sep="\t", index_col=0)
year_counts = pd.read_csv(COUNT_FILE, sep="\t")
pos_map = pd.read_csv(POSMAP_FILE, sep="\t")

aa_mat.index = aa_mat.index.astype(int)
percent_mat.index = percent_mat.index.astype(int)

position_cols = aa_mat.columns.tolist()
years = aa_mat.index.tolist()

print("年份数:", len(years))
print("位点数:", len(position_cols))

display(long_df.head())
display(aa_mat.head())
display(percent_mat.head())
display(year_counts.head())


# ============================================================
# 11. 美化版：每年序列数量柱状图
# ============================================================

FIG_W = 18
FIG_H = 12

TITLE_FONTSIZE = 20
LABEL_FONTSIZE = 16
TICK_FONTSIZE = 12
ANNOT_FONTSIZE = 10

BAR_COLOR = "#3B82C4"
EDGE_COLOR = "black"
EDGE_WIDTH = 0.4

SHOW_GRID = False
GRID_ALPHA = 0.25

Y_TOP_MARGIN = 0.12

ROTATION = 45

SAVE_DPI = 600

plot_df = year_counts.copy()
plot_df = plot_df.sort_values("year")

plt.figure(figsize=(FIG_W, FIG_H))

ax = sns.barplot(
    data=plot_df,
    x="year",
    y="n_sequences",
    color=BAR_COLOR,
    edgecolor=EDGE_COLOR,
    linewidth=EDGE_WIDTH
)

for patch in ax.patches:
    height = patch.get_height()

    if np.isnan(height):
        continue

    x = patch.get_x() + patch.get_width() / 2

    ax.text(
        x,
        height,
        f"{int(height)}",
        ha="center",
        va="bottom",
        fontsize=ANNOT_FONTSIZE,
        color="black"
    )

ax.set_title(
    "Number of HA sequences per year",
    fontsize=TITLE_FONTSIZE,
    fontweight="bold",
    pad=15
)

ax.set_xlabel(
    "Year",
    fontsize=LABEL_FONTSIZE,
    labelpad=10
)

ax.set_ylabel(
    "Number of sequences",
    fontsize=LABEL_FONTSIZE,
    labelpad=10
)

ax.tick_params(axis="x", labelsize=TICK_FONTSIZE, rotation=ROTATION)
ax.tick_params(axis="y", labelsize=TICK_FONTSIZE)

ymax = plot_df["n_sequences"].max()
ax.set_ylim(0, ymax * (1 + Y_TOP_MARGIN))

if SHOW_GRID:
    ax.yaxis.grid(True, linestyle="--", alpha=GRID_ALPHA)
    ax.xaxis.grid(False)
else:
    ax.grid(False)

sns.despine(ax=ax)

plt.tight_layout()

png_file = os.path.join(FIGDIR, "yearly_sequence_counts_beautified.png")
pdf_file = os.path.join(FIGDIR, "yearly_sequence_counts_beautified.pdf")

plt.savefig(png_file, dpi=SAVE_DPI, bbox_inches="tight")
plt.savefig(pdf_file, bbox_inches="tight")

plt.show()

print("Figure saved to:")
print(png_file)
print(pdf_file)
```

```python
import re
import os
import numpy as np
import pandas as pd
import seaborn as sns

from openpyxl.styles import PatternFill, Font, Alignment, Border, Side
from openpyxl.utils import get_column_letter


# ============================================================
# 根据指定 AA 位点导出带颜色的 Excel heatmap
# cell 内显示氨基酸缩写
# ============================================================


# ============================================================
# 0. AA 固定配色
# 与当前 Amino acid 图例颜色保持一致
# ============================================================

AA_ORDER_SELECTED = [
    "A", "C", "D", "E", "F", "G",
    "H", "I", "K", "L", "M", "N",
    "Q", "R", "S", "T", "V", "Y"
]

AA_COLOR_DICT_SELECTED = {
    "A": "#1f77b4",
    "C": "#aec7e8",
    "D": "#ff7f0e",
    "E": "#ffbb78",
    "F": "#2ca02c",
    "G": "#98df8a",
    "H": "#d62728",
    "I": "#ff9896",
    "K": "#9467bd",
    "L": "#c5b0d5",
    "M": "#8c564b",
    "N": "#c49c94",
    "Q": "#e377c2",
    "R": "#f7b6d2",
    "S": "#7f7f7f",
    "T": "#c7c7c7",
    "V": "#bcbd22",
    "Y": "#dbdb8d",
}


# ============================================================
# 1. 指定要导出的 AA 位点
# ============================================================

# 支持以下格式：
# 145
# "145"
# "pos_145"
# "HA_145"
SELECTED_AA_SITES = [
  66,69,70,71,73,74,75,76,78,79,94,95,98,106,108,289,290,291,292,293,294
]

# 是否按照 position 数字排序
SORT_SELECTED_SITES = True

# 输出文件名前缀
SELECTED_FIG_PREFIX = "selected_AA_state_heatmap_with_AA_text"


# ============================================================
# 2. Excel 输出参数
# ============================================================

EXCEL_FILENAME_SELECTED = f"{SELECTED_FIG_PREFIX}_colored_excel_heatmap.xlsx"

# Excel 中缺失值显示内容
NA_TEXT_SELECTED = ""

# Excel 中缺失值背景色
NA_COLOR_SELECTED = "FFFFFF"

# 是否根据背景色自动设置文字颜色
# True: 深色背景用白字，浅色背景用黑字
# False: 使用固定颜色 AA_TEXT_COLOR_SELECTED
AA_TEXT_AUTO_COLOR_SELECTED = False

# 如果 AA_TEXT_AUTO_COLOR_SELECTED = False，则使用这个固定颜色
AA_TEXT_COLOR_SELECTED = "000000"

# 自动黑白文字的亮度阈值
AA_TEXT_LUMINANCE_THRESHOLD_SELECTED = 0.55

# AA 文字字体
AA_TEXT_FONT_NAME_SELECTED = "Arial"

# AA 文字字号
AA_TEXT_FONTSIZE_SELECTED = 14

# AA 文字是否加粗
AA_TEXT_BOLD_SELECTED = True

# 表头字体
HEADER_FONT_NAME_SELECTED = "Arial"
HEADER_FONTSIZE_SELECTED = 12
HEADER_BOLD_SELECTED = True

# Year 行名字体
INDEX_FONT_NAME_SELECTED = "Arial"
INDEX_FONTSIZE_SELECTED = 12
INDEX_BOLD_SELECTED = True

# 单元格大小
EXCEL_CELL_WIDTH_SELECTED = 6
EXCEL_CELL_HEIGHT_SELECTED = 24

# Year 列宽度
EXCEL_YEAR_COLUMN_WIDTH_SELECTED = 12

# 是否冻结首行首列
FREEZE_PANES_SELECTED = True

# 网格线 / 边框
SHOW_CELL_BORDER_SELECTED = True
BORDER_COLOR_SELECTED = "D3D3D3"
BORDER_STYLE_SELECTED = "thin"

# 数字编码中缺失值编码
NA_CODE_SELECTED = -1


# ============================================================
# 工具函数
# ============================================================

def get_pos_number_selected(col):
    """
    从列名或输入位点中提取数字 position。
    支持：
    pos_123
    123
    HA_123
    """
    match = re.search(r'\d+', str(col))
    if match:
        return int(match.group())
    else:
        return float("inf")


def normalize_site_to_pos_selected(site):
    """
    把用户输入的位点转换成数字 position。
    支持:
    145
    "145"
    "pos_145"
    "HA_145"
    """
    match = re.search(r'\d+', str(site))
    if match:
        return int(match.group())
    else:
        raise ValueError(f"Cannot parse AA site: {site}")


def hex_to_excel_rgb_selected(hex_color):
    """
    把 #RRGGBB 转成 openpyxl 使用的 RRGGBB。
    """
    if pd.isna(hex_color):
        return "FFFFFF"

    hex_color = str(hex_color).strip()

    if hex_color.startswith("#"):
        hex_color = hex_color[1:]

    hex_color = hex_color.upper()

    if len(hex_color) == 6:
        return hex_color
    else:
        return "FFFFFF"


def choose_text_color_from_hex_selected(hex_color, threshold=0.55):
    """
    根据背景色判断 Excel 文字使用黑色还是白色。
    返回 RRGGBB 格式。
    """
    hex_color = hex_to_excel_rgb_selected(hex_color)

    r = int(hex_color[0:2], 16) / 255
    g = int(hex_color[2:4], 16) / 255
    b = int(hex_color[4:6], 16) / 255

    luminance = 0.299 * r + 0.587 * g + 0.114 * b

    if luminance > threshold:
        return "000000"
    else:
        return "FFFFFF"


# ============================================================
# 建立 position number -> aa_mat column name 映射
# ============================================================

pos_to_col_selected = {}

for col in aa_mat.columns:
    pos = get_pos_number_selected(col)
    if pos != float("inf"):
        pos_to_col_selected[pos] = col


# ============================================================
# 处理指定 AA 位点
# ============================================================

selected_positions = [
    normalize_site_to_pos_selected(site)
    for site in SELECTED_AA_SITES
]

if SORT_SELECTED_SITES:
    selected_positions = sorted(selected_positions)


# 检查重复输入位点
duplicated_positions_series = pd.Series(selected_positions)
duplicated_positions = duplicated_positions_series[
    duplicated_positions_series.duplicated()
].tolist()

if len(duplicated_positions) > 0:
    print("Warning: duplicated positions in SELECTED_AA_SITES:")
    print(duplicated_positions)


# 检查缺失位点
missing_positions = [
    pos for pos in selected_positions
    if pos not in pos_to_col_selected
]

if len(missing_positions) > 0:
    print("Warning: these positions are not found in aa_mat:")
    print(missing_positions)


# 只保留 aa_mat 中存在的位点
selected_positions = [
    pos for pos in selected_positions
    if pos in pos_to_col_selected
]

selected_cols = [
    pos_to_col_selected[pos]
    for pos in selected_positions
]

if len(selected_cols) == 0:
    raise ValueError("No selected AA sites were found in aa_mat.")


print("Selected AA positions:")
print(selected_positions)

print("Selected columns:")
print(selected_cols)


# ============================================================
# 构建选定位点矩阵
# ============================================================

selected_aa_mat = aa_mat[selected_cols].copy()

# Excel 中列名直接显示 position 数字
selected_aa_mat.columns = [
    str(pos) for pos in selected_positions
]


# ============================================================
# 为了保持颜色一致，使用整个 aa_mat 中出现过的 AA 建立颜色映射
# 使用指定 AA 配色，顺序与当前 Amino acid 图例一致
# ============================================================

all_sites_selected = list(aa_mat.columns)
all_sites_selected = sorted(
    all_sites_selected,
    key=get_pos_number_selected
)

all_aa_mat_selected = aa_mat[all_sites_selected].copy()

all_aas_raw_selected = pd.unique(
    all_aa_mat_selected.values.ravel()
)

all_aas_raw_selected = [
    a for a in all_aas_raw_selected
    if pd.notna(a)
]

# 先按照指定 AA_ORDER_SELECTED 排列，再追加不在指定顺序中的其他 AA
all_aas_selected = [
    aa for aa in AA_ORDER_SELECTED
    if aa in all_aas_raw_selected
]

extra_aas_selected = [
    aa for aa in all_aas_raw_selected
    if aa not in AA_ORDER_SELECTED
]

extra_aas_selected = sorted(extra_aas_selected)

all_aas_selected = all_aas_selected + extra_aas_selected

print("AA states:")
print(all_aas_selected)


aa_to_int_selected = {
    aa: i for i, aa in enumerate(all_aas_selected)
}

int_to_aa_selected = {
    i: aa for aa, i in aa_to_int_selected.items()
}


encoded_selected_mat = selected_aa_mat.replace(
    aa_to_int_selected
)

encoded_selected_mat_for_excel = encoded_selected_mat.fillna(
    NA_CODE_SELECTED
).astype(int)


# ============================================================
# 颜色表
# 未在 AA_COLOR_DICT_SELECTED 中的 AA 使用 tab20 备用色
# ============================================================

fallback_palette_selected = sns.color_palette(
    "tab20",
    n_colors=max(20, len(all_aas_selected))
).as_hex()

aa_to_color_selected = {}

for i, aa in enumerate(all_aas_selected):
    if aa in AA_COLOR_DICT_SELECTED:
        aa_to_color_selected[aa] = AA_COLOR_DICT_SELECTED[aa]
    else:
        fallback_color = fallback_palette_selected[i % len(fallback_palette_selected)]
        aa_to_color_selected[aa] = fallback_color
        print(
            f"Warning: {aa} not found in AA_COLOR_DICT_SELECTED, "
            f"using fallback color {fallback_color}."
        )


# ============================================================
# 构建输出表
# ============================================================

# 1. 带 AA 字母的矩阵
colored_aa_df = selected_aa_mat.copy()
colored_aa_df = colored_aa_df.fillna(NA_TEXT_SELECTED)
colored_aa_df.insert(
    0,
    "Year",
    colored_aa_df.index
)

# 2. 数字编码矩阵
numeric_code_df = encoded_selected_mat_for_excel.copy()
numeric_code_df.insert(
    0,
    "Year",
    numeric_code_df.index
)

# 3. AA 编码和颜色表
aa_code_color_df = pd.DataFrame({
    "Amino_acid": all_aas_selected,
    "Numeric_code": [
        aa_to_int_selected[aa] for aa in all_aas_selected
    ],
    "Hex_color": [
        aa_to_color_selected[aa] for aa in all_aas_selected
    ]
})

aa_code_color_df = pd.concat(
    [
        pd.DataFrame({
            "Amino_acid": ["NA"],
            "Numeric_code": [NA_CODE_SELECTED],
            "Hex_color": [f"#{NA_COLOR_SELECTED}"]
        }),
        aa_code_color_df
    ],
    ignore_index=True
)


# 4. Long format
long_records = []

for year in selected_aa_mat.index:
    for pos in selected_aa_mat.columns:

        aa_value = selected_aa_mat.loc[year, pos]

        if pd.isna(aa_value):
            aa_text = "NA"
            code_value = NA_CODE_SELECTED
            color_value = f"#{NA_COLOR_SELECTED}"
        else:
            aa_text = str(aa_value)
            code_value = aa_to_int_selected[aa_text]
            color_value = aa_to_color_selected[aa_text]

        long_records.append({
            "Year": year,
            "HA_position": pos,
            "Amino_acid": aa_text,
            "Numeric_code": code_value,
            "Hex_color": color_value
        })

long_format_df = pd.DataFrame(long_records)


# ============================================================
# 输出 Excel
# ============================================================

excel_outpath = os.path.join(
    FIGDIR,
    EXCEL_FILENAME_SELECTED
)

with pd.ExcelWriter(excel_outpath, engine="openpyxl") as writer:

    colored_aa_df.to_excel(
        writer,
        sheet_name="Colored_AA_heatmap",
        index=False
    )

    numeric_code_df.to_excel(
        writer,
        sheet_name="Numeric_code_heatmap",
        index=False
    )

    aa_code_color_df.to_excel(
        writer,
        sheet_name="AA_code_color_legend",
        index=False
    )

    long_format_df.to_excel(
        writer,
        sheet_name="Long_format",
        index=False
    )

    workbook = writer.book

    # ========================================================
    # 格式化 Colored_AA_heatmap
    # ========================================================

    ws = workbook["Colored_AA_heatmap"]

    max_row = ws.max_row
    max_col = ws.max_column

    # 冻结首行和首列
    if FREEZE_PANES_SELECTED:
        ws.freeze_panes = "B2"

    # 是否显示 Excel 默认网格线
    ws.sheet_view.showGridLines = False

    # 边框
    if SHOW_CELL_BORDER_SELECTED:
        side = Side(
            style=BORDER_STYLE_SELECTED,
            color=BORDER_COLOR_SELECTED
        )
        cell_border = Border(
            left=side,
            right=side,
            top=side,
            bottom=side
        )
    else:
        cell_border = Border()

    # 表头格式
    header_fill = PatternFill(
        fill_type="solid",
        fgColor="F2F2F2"
    )

    for col_idx in range(1, max_col + 1):
        cell = ws.cell(row=1, column=col_idx)
        cell.fill = header_fill
        cell.font = Font(
            name=HEADER_FONT_NAME_SELECTED,
            size=HEADER_FONTSIZE_SELECTED,
            bold=HEADER_BOLD_SELECTED,
            color="000000"
        )
        cell.alignment = Alignment(
            horizontal="center",
            vertical="center"
        )
        cell.border = cell_border

    # Year 列格式
    for row_idx in range(2, max_row + 1):
        cell = ws.cell(row=row_idx, column=1)
        cell.fill = header_fill
        cell.font = Font(
            name=INDEX_FONT_NAME_SELECTED,
            size=INDEX_FONTSIZE_SELECTED,
            bold=INDEX_BOLD_SELECTED,
            color="000000"
        )
        cell.alignment = Alignment(
            horizontal="center",
            vertical="center"
        )
        cell.border = cell_border

    # AA heatmap cell 着色
    for row_idx in range(2, max_row + 1):
        for col_idx in range(2, max_col + 1):

            cell = ws.cell(row=row_idx, column=col_idx)
            aa_value = cell.value

            if aa_value is None or str(aa_value) == "":
                bg_color = NA_COLOR_SELECTED
                text_color = "000000"
                cell.value = NA_TEXT_SELECTED
            else:
                aa_value = str(aa_value)

                if aa_value in aa_to_color_selected:
                    bg_color = hex_to_excel_rgb_selected(
                        aa_to_color_selected[aa_value]
                    )
                else:
                    bg_color = NA_COLOR_SELECTED

                if AA_TEXT_AUTO_COLOR_SELECTED:
                    text_color = choose_text_color_from_hex_selected(
                        bg_color,
                        threshold=AA_TEXT_LUMINANCE_THRESHOLD_SELECTED
                    )
                else:
                    text_color = hex_to_excel_rgb_selected(
                        AA_TEXT_COLOR_SELECTED
                    )

            cell.fill = PatternFill(
                fill_type="solid",
                fgColor=bg_color
            )

            cell.font = Font(
                name=AA_TEXT_FONT_NAME_SELECTED,
                size=AA_TEXT_FONTSIZE_SELECTED,
                bold=AA_TEXT_BOLD_SELECTED,
                color=text_color
            )

            cell.alignment = Alignment(
                horizontal="center",
                vertical="center"
            )

            cell.border = cell_border

    # 设置列宽
    ws.column_dimensions["A"].width = EXCEL_YEAR_COLUMN_WIDTH_SELECTED

    for col_idx in range(2, max_col + 1):
        col_letter = get_column_letter(col_idx)
        ws.column_dimensions[col_letter].width = EXCEL_CELL_WIDTH_SELECTED

    # 设置行高
    for row_idx in range(1, max_row + 1):
        ws.row_dimensions[row_idx].height = EXCEL_CELL_HEIGHT_SELECTED

    # ========================================================
    # 格式化 Numeric_code_heatmap
    # ========================================================

    ws_num = workbook["Numeric_code_heatmap"]

    ws_num.freeze_panes = "B2"
    ws_num.sheet_view.showGridLines = False

    for row in ws_num.iter_rows():
        for cell in row:
            cell.alignment = Alignment(
                horizontal="center",
                vertical="center"
            )
            cell.border = cell_border
            cell.font = Font(
                name="Arial",
                size=11,
                color="000000"
            )

    for cell in ws_num[1]:
        cell.fill = header_fill
        cell.font = Font(
            name="Arial",
            size=11,
            bold=True,
            color="000000"
        )

    for row_idx in range(2, ws_num.max_row + 1):
        ws_num.cell(row=row_idx, column=1).fill = header_fill
        ws_num.cell(row=row_idx, column=1).font = Font(
            name="Arial",
            size=11,
            bold=True,
            color="000000"
        )

    ws_num.column_dimensions["A"].width = EXCEL_YEAR_COLUMN_WIDTH_SELECTED

    for col_idx in range(2, ws_num.max_column + 1):
        ws_num.column_dimensions[get_column_letter(col_idx)].width = EXCEL_CELL_WIDTH_SELECTED

    for row_idx in range(1, ws_num.max_row + 1):
        ws_num.row_dimensions[row_idx].height = EXCEL_CELL_HEIGHT_SELECTED

    # ========================================================
    # 格式化 AA_code_color_legend
    # ========================================================

    ws_legend = workbook["AA_code_color_legend"]

    ws_legend.sheet_view.showGridLines = False

    for row in ws_legend.iter_rows():
        for cell in row:
            cell.alignment = Alignment(
                horizontal="center",
                vertical="center"
            )
            cell.border = cell_border
            cell.font = Font(
                name="Arial",
                size=11,
                color="000000"
            )

    for cell in ws_legend[1]:
        cell.fill = header_fill
        cell.font = Font(
            name="Arial",
            size=11,
            bold=True,
            color="000000"
        )

    # 给 legend 的 Hex_color 对应行上色
    # 列位置：
    # A = Amino_acid
    # B = Numeric_code
    # C = Hex_color
    for row_idx in range(2, ws_legend.max_row + 1):

        aa_cell = ws_legend.cell(row=row_idx, column=1)
        code_cell = ws_legend.cell(row=row_idx, column=2)
        color_cell = ws_legend.cell(row=row_idx, column=3)

        hex_color = color_cell.value
        bg_color = hex_to_excel_rgb_selected(hex_color)

        for cell in [aa_cell, code_cell, color_cell]:
            cell.fill = PatternFill(
                fill_type="solid",
                fgColor=bg_color
            )

            if AA_TEXT_AUTO_COLOR_SELECTED:
                text_color = choose_text_color_from_hex_selected(
                    bg_color,
                    threshold=AA_TEXT_LUMINANCE_THRESHOLD_SELECTED
                )
            else:
                text_color = "000000"

            cell.font = Font(
                name="Arial",
                size=11,
                bold=True,
                color=text_color
            )

    ws_legend.column_dimensions["A"].width = 15
    ws_legend.column_dimensions["B"].width = 18
    ws_legend.column_dimensions["C"].width = 15

    for row_idx in range(1, ws_legend.max_row + 1):
        ws_legend.row_dimensions[row_idx].height = 22

    # ========================================================
    # 格式化 Long_format
    # ========================================================

    ws_long = workbook["Long_format"]
    ws_long.sheet_view.showGridLines = False

    for row in ws_long.iter_rows():
        for cell in row:
            cell.alignment = Alignment(
                horizontal="center",
                vertical="center"
            )
            cell.border = cell_border
            cell.font = Font(
                name="Arial",
                size=11,
                color="000000"
            )

    for cell in ws_long[1]:
        cell.fill = header_fill
        cell.font = Font(
            name="Arial",
            size=11,
            bold=True,
            color="000000"
        )

    for col_idx in range(1, ws_long.max_column + 1):
        ws_long.column_dimensions[get_column_letter(col_idx)].width = 16


print(f"Saved colored Excel heatmap: {excel_outpath}")


# ============================================================
# 预览信息
# ============================================================

print("\nColored AA heatmap preview:")
print(colored_aa_df.head())

print("\nAA code color legend:")
print(aa_code_color_df)
```

```python
# ============================================================
# Jupyter / ipynb 直接输出版
# 按年份导出 fasta，并统计每年每个位点 AA 百分比
# ============================================================

import os
import re
import zipfile
import pandas as pd
from collections import Counter

from IPython.display import display, HTML, FileLink

from openpyxl.styles import PatternFill, Font, Alignment, Border, Side
from openpyxl.utils import get_column_letter


# ============================================================
# 0. 基础检查与输出目录
# ============================================================

# 如果你前面已经定义了 FIGDIR，就用已有 FIGDIR
# 如果没有定义，则自动创建当前目录下的 figures
try:
    FIGDIR
except NameError:
    FIGDIR = "figures"

os.makedirs(FIGDIR, exist_ok=True)

# 你的原始序列表
# 当前列名：
# ['accession', 'year', 'header', 'sequence', 'length']
SEQ_DF = seq_df.copy()

SEQ_ID_COL = "accession"
YEAR_COL = "year"
HEADER_COL = "header"
SEQ_COL = "sequence"

# 需要统计的 AA 位点
SELECTED_AA_SITES = [
    66, 69, 70, 71, 73, 74, 75, 76, 78, 79,
    94, 95, 98, 106, 108,
    289, 290, 291, 292, 293, 294
]

# True: 只统计上面指定的位点
# False: 统计整条序列所有位置
USE_SELECTED_SITES_ONLY = True

# 是否按照位点数字排序
SORT_SELECTED_SITES = True

# 默认 HA position 是 1-based
# position 66 对应 sequence[65]
POSITION_IS_ONE_BASED = True

# 输出目录
FASTA_BY_YEAR_DIR = os.path.join(FIGDIR, "fasta_by_year")
SITE_PERCENT_DIR = os.path.join(FIGDIR, "site_percentage_by_year")

os.makedirs(FASTA_BY_YEAR_DIR, exist_ok=True)
os.makedirs(SITE_PERCENT_DIR, exist_ok=True)

# 输出文件
FASTA_PREFIX = "sequences"

SITE_PERCENT_EXCEL = os.path.join(
    SITE_PERCENT_DIR,
    "AA_site_percentage_by_year.xlsx"
)

SITE_PERCENT_LONG_CSV = os.path.join(
    SITE_PERCENT_DIR,
    "AA_site_percentage_by_year_long.csv"
)

SITE_PERCENT_WIDE_CSV = os.path.join(
    SITE_PERCENT_DIR,
    "AA_site_percentage_by_year_wide.csv"
)

DOMINANT_AA_CSV = os.path.join(
    SITE_PERCENT_DIR,
    "dominant_AA_by_year_site.csv"
)

FASTA_SUMMARY_CSV = os.path.join(
    FASTA_BY_YEAR_DIR,
    "fasta_by_year_summary.csv"
)

FASTA_ZIP_PATH = os.path.join(
    FIGDIR,
    "fasta_by_year.zip"
)


# ============================================================
# 1. 工具函数
# ============================================================

def normalize_site_to_pos(site):
    """
    把输入位点转换成数字 position。
    支持：
    145
    "145"
    "pos_145"
    "HA_145"
    """
    match = re.search(r"\d+", str(site))
    if match:
        return int(match.group())
    else:
        raise ValueError(f"Cannot parse AA site: {site}")


def sanitize_filename(text):
    """
    清理文件名，避免年份或其他字段中的特殊字符影响保存。
    """
    text = str(text)
    text = re.sub(r"[^\w\-\.]+", "_", text)
    return text


def wrap_fasta_sequence(seq, width=80):
    """
    fasta 序列自动换行。
    """
    seq = str(seq)
    return "\n".join(
        seq[i:i + width]
        for i in range(0, len(seq), width)
    )


def write_fasta_for_one_year(df_year, year, outpath):
    """
    写出某一年份的 fasta 文件。
    """
    with open(outpath, "w") as f:
        for _, row in df_year.iterrows():

            seq_id = str(row[SEQ_ID_COL])
            seq = str(row[SEQ_COL])
            year_text = str(row[YEAR_COL])

            if HEADER_COL in row.index and pd.notna(row[HEADER_COL]):
                original_header = str(row[HEADER_COL])
                original_header = original_header.replace("\n", " ").replace("\r", " ")
                fasta_header = f">{seq_id}|Year={year_text}|{original_header}"
            else:
                fasta_header = f">{seq_id}|Year={year_text}"

            f.write(fasta_header + "\n")
            f.write(wrap_fasta_sequence(seq) + "\n")


def make_zip_from_folder(folder_path, zip_path):
    """
    把一个文件夹压缩成 zip。
    """
    with zipfile.ZipFile(zip_path, "w", zipfile.ZIP_DEFLATED) as zipf:
        for root, dirs, files in os.walk(folder_path):
            for file in files:
                file_path = os.path.join(root, file)
                arcname = os.path.relpath(file_path, folder_path)
                zipf.write(file_path, arcname)


def display_section(title):
    display(HTML(f"""
    <h2 style="
        color:#1f4e79;
        border-bottom:2px solid #1f4e79;
        padding-bottom:4px;
        margin-top:25px;
    ">
    {title}
    </h2>
    """))


# ============================================================
# 2. 检查输入数据
# ============================================================

display_section("1. Input data check")

required_cols = [
    SEQ_ID_COL,
    YEAR_COL,
    SEQ_COL
]

missing_cols = [
    col for col in required_cols
    if col not in SEQ_DF.columns
]

if len(missing_cols) > 0:
    raise ValueError(
        f"SEQ_DF is missing required columns: {missing_cols}\n"
        f"Current columns are: {list(SEQ_DF.columns)}"
    )

input_info_df = pd.DataFrame({
    "Item": [
        "Input dataframe",
        "Original records",
        "Columns",
        "Sequence ID column",
        "Year column",
        "Sequence column"
    ],
    "Value": [
        "seq_df",
        len(SEQ_DF),
        ", ".join(list(SEQ_DF.columns)),
        SEQ_ID_COL,
        YEAR_COL,
        SEQ_COL
    ]
})

display(input_info_df)


# ============================================================
# 3. 清理 year 和 sequence
# ============================================================

display_section("2. Clean sequence table")

# 去掉没有年份或没有序列的记录
SEQ_DF = SEQ_DF.dropna(subset=[YEAR_COL, SEQ_COL]).copy()

# year 转成整数年份
SEQ_DF[YEAR_COL] = pd.to_numeric(
    SEQ_DF[YEAR_COL],
    errors="coerce"
)

SEQ_DF = SEQ_DF.dropna(subset=[YEAR_COL]).copy()
SEQ_DF[YEAR_COL] = SEQ_DF[YEAR_COL].astype(int)

# sequence 清理
SEQ_DF[SEQ_COL] = (
    SEQ_DF[SEQ_COL]
    .astype(str)
    .str.upper()
    .str.replace(" ", "", regex=False)
    .str.replace("\n", "", regex=False)
    .str.replace("\r", "", regex=False)
    .str.replace("\t", "", regex=False)
)

# 去掉空序列
SEQ_DF = SEQ_DF[SEQ_DF[SEQ_COL].str.len() > 0].copy()

# accession 缺失时自动生成
SEQ_DF[SEQ_ID_COL] = SEQ_DF[SEQ_ID_COL].astype(str)

missing_accession_mask = (
    SEQ_DF[SEQ_ID_COL].isna()
    | (SEQ_DF[SEQ_ID_COL].astype(str).str.strip() == "")
    | (SEQ_DF[SEQ_ID_COL].astype(str).str.lower() == "nan")
)

SEQ_DF.loc[missing_accession_mask, SEQ_ID_COL] = [
    f"seq_{i}"
    for i in SEQ_DF.index[missing_accession_mask]
]

# 更新 clean_length
SEQ_DF["clean_length"] = SEQ_DF[SEQ_COL].str.len()

clean_summary_df = pd.DataFrame({
    "Item": [
        "Cleaned sequences",
        "Number of years",
        "Min year",
        "Max year",
        "Min length",
        "Max length",
        "Mean length"
    ],
    "Value": [
        len(SEQ_DF),
        SEQ_DF[YEAR_COL].nunique(),
        SEQ_DF[YEAR_COL].min(),
        SEQ_DF[YEAR_COL].max(),
        SEQ_DF["clean_length"].min(),
        SEQ_DF["clean_length"].max(),
        round(SEQ_DF["clean_length"].mean(), 2)
    ]
})

display(clean_summary_df)

display(HTML("<b>Cleaned sequence preview:</b>"))
display(SEQ_DF.head())


# ============================================================
# 4. 按年份输出 fasta
# ============================================================

display_section("3. Export fasta files by year")

fasta_summary_records = []

for year, df_year in SEQ_DF.groupby(YEAR_COL):

    year_safe = sanitize_filename(year)

    fasta_outpath = os.path.join(
        FASTA_BY_YEAR_DIR,
        f"{FASTA_PREFIX}_{year_safe}.fasta"
    )

    write_fasta_for_one_year(
        df_year=df_year,
        year=year,
        outpath=fasta_outpath
    )

    fasta_summary_records.append({
        "Year": year,
        "N_sequences": len(df_year),
        "Min_length": df_year["clean_length"].min(),
        "Max_length": df_year["clean_length"].max(),
        "Mean_length": round(df_year["clean_length"].mean(), 2),
        "Fasta_file": fasta_outpath
    })

fasta_summary_df = pd.DataFrame(fasta_summary_records)

fasta_summary_df = fasta_summary_df.sort_values(
    by="Year"
).reset_index(drop=True)

fasta_summary_df.to_csv(
    FASTA_SUMMARY_CSV,
    index=False
)

display(fasta_summary_df)


# ============================================================
# 5. 确定需要统计的位点
# ============================================================

display_section("4. Positions used for percentage calculation")

if USE_SELECTED_SITES_ONLY:

    positions_for_percentage = [
        normalize_site_to_pos(site)
        for site in SELECTED_AA_SITES
    ]

    if SORT_SELECTED_SITES:
        positions_for_percentage = sorted(positions_for_percentage)

else:
    max_seq_len = SEQ_DF[SEQ_COL].str.len().max()
    positions_for_percentage = list(range(1, max_seq_len + 1))

position_df = pd.DataFrame({
    "HA_position": positions_for_percentage
})

display(position_df)


# ============================================================
# 6. 统计每个年份、每个位点 AA 数量和百分比
# ============================================================

display_section("5. AA percentage by year and position")

percentage_records = []

for year, df_year in SEQ_DF.groupby(YEAR_COL):

    seqs = df_year[SEQ_COL].tolist()
    n_seq_total = len(seqs)

    for pos in positions_for_percentage:

        if POSITION_IS_ONE_BASED:
            seq_index = pos - 1
        else:
            seq_index = pos

        aa_list = []

        for seq in seqs:

            if seq_index < 0:
                aa = "NA"
            elif seq_index >= len(seq):
                aa = "NA"
            else:
                aa = seq[seq_index]

            if pd.isna(aa) or str(aa).strip() == "":
                aa = "NA"
            else:
                aa = str(aa).upper()

            aa_list.append(aa)

        counter = Counter(aa_list)

        for aa, count in sorted(counter.items()):

            percentage = (
                count / n_seq_total * 100
                if n_seq_total > 0
                else 0
            )

            percentage_records.append({
                "Year": year,
                "HA_position": pos,
                "Amino_acid": aa,
                "Count": count,
                "Total_sequences": n_seq_total,
                "Percentage": percentage
            })

site_percentage_df = pd.DataFrame(percentage_records)

site_percentage_df = site_percentage_df.sort_values(
    by=["Year", "HA_position", "Amino_acid"]
).reset_index(drop=True)

display(HTML("<b>Long format preview:</b>"))
display(site_percentage_df.head(30))


# ============================================================
# 7. Wide format
# ============================================================

display_section("6. AA percentage wide format")

site_percentage_wide_df = site_percentage_df.pivot_table(
    index=["Year", "HA_position"],
    columns="Amino_acid",
    values="Percentage",
    fill_value=0
).reset_index()

site_percentage_wide_df.columns.name = None

# AA 列排序
AA_ORDER_SELECTED = [
    "A", "C", "D", "E", "F", "G",
    "H", "I", "K", "L", "M", "N",
    "Q", "R", "S", "T", "V", "Y"
]

base_cols = ["Year", "HA_position"]

aa_cols = [
    col for col in site_percentage_wide_df.columns
    if col not in base_cols
]

aa_cols_ordered = [
    aa for aa in AA_ORDER_SELECTED
    if aa in aa_cols
]

extra_aa_cols = sorted([
    aa for aa in aa_cols
    if aa not in aa_cols_ordered
])

site_percentage_wide_df = site_percentage_wide_df[
    base_cols + aa_cols_ordered + extra_aa_cols
]

display(site_percentage_wide_df.head(30))


# ============================================================
# 8. Dominant AA
# ============================================================

display_section("7. Dominant AA by year and position")

dominant_aa_df = (
    site_percentage_df
    .sort_values(
        by=["Year", "HA_position", "Percentage", "Count", "Amino_acid"],
        ascending=[True, True, False, False, True]
    )
    .groupby(["Year", "HA_position"], as_index=False)
    .first()
)

dominant_aa_df = dominant_aa_df.rename(columns={
    "Amino_acid": "Dominant_AA",
    "Count": "Dominant_count",
    "Percentage": "Dominant_percentage"
})

dominant_aa_df = dominant_aa_df[
    [
        "Year",
        "HA_position",
        "Dominant_AA",
        "Dominant_count",
        "Total_sequences",
        "Dominant_percentage"
    ]
]

display(dominant_aa_df.head(30))


# ============================================================
# 9. 保存 CSV
# ============================================================

site_percentage_df.to_csv(
    SITE_PERCENT_LONG_CSV,
    index=False
)

site_percentage_wide_df.to_csv(
    SITE_PERCENT_WIDE_CSV,
    index=False
)

dominant_aa_df.to_csv(
    DOMINANT_AA_CSV,
    index=False
)


# ============================================================
# 10. 保存 Excel
# ============================================================

display_section("8. Save Excel result")

with pd.ExcelWriter(SITE_PERCENT_EXCEL, engine="openpyxl") as writer:

    fasta_summary_df.to_excel(
        writer,
        sheet_name="Fasta_by_year_summary",
        index=False
    )

    site_percentage_df.to_excel(
        writer,
        sheet_name="AA_percentage_long",
        index=False
    )

    site_percentage_wide_df.to_excel(
        writer,
        sheet_name="AA_percentage_wide",
        index=False
    )

    dominant_aa_df.to_excel(
        writer,
        sheet_name="Dominant_AA",
        index=False
    )

    workbook = writer.book

    header_fill = PatternFill(
        fill_type="solid",
        fgColor="F2F2F2"
    )

    side = Side(
        style="thin",
        color="D3D3D3"
    )

    cell_border = Border(
        left=side,
        right=side,
        top=side,
        bottom=side
    )

    for sheet_name in workbook.sheetnames:

        ws = workbook[sheet_name]
        ws.sheet_view.showGridLines = False
        ws.freeze_panes = "A2"

        for row in ws.iter_rows():
            for cell in row:
                cell.alignment = Alignment(
                    horizontal="center",
                    vertical="center"
                )
                cell.border = cell_border
                cell.font = Font(
                    name="Arial",
                    size=11,
                    color="000000"
                )

        for cell in ws[1]:
            cell.fill = header_fill
            cell.font = Font(
                name="Arial",
                size=11,
                bold=True,
                color="000000"
            )

        for col_idx in range(1, ws.max_column + 1):

            col_letter = get_column_letter(col_idx)
            max_length = 0

            for row_idx in range(1, ws.max_row + 1):
                value = ws.cell(row=row_idx, column=col_idx).value
                if value is not None:
                    max_length = max(max_length, len(str(value)))

            adjusted_width = min(max(max_length + 2, 10), 60)
            ws.column_dimensions[col_letter].width = adjusted_width

        for row_idx in range(1, ws.max_row + 1):
            ws.row_dimensions[row_idx].height = 20

display(HTML(f"""
<p style="color:green; font-weight:bold;">
Excel saved successfully:
</p>
<code>{SITE_PERCENT_EXCEL}</code>
"""))


# ============================================================
# 11. 打包 fasta 文件夹
# ============================================================

display_section("9. Zip fasta files")

make_zip_from_folder(
    folder_path=FASTA_BY_YEAR_DIR,
    zip_path=FASTA_ZIP_PATH
)

display(HTML(f"""
<p style="color:green; font-weight:bold;">
Fasta files zipped successfully:
</p>
<code>{FASTA_ZIP_PATH}</code>
"""))


# ============================================================
# 12. ipynb 中直接显示下载链接
# ============================================================

display_section("10. Output files")

output_files_df = pd.DataFrame({
    "File_type": [
        "Fasta zip",
        "Fasta summary CSV",
        "AA percentage long CSV",
        "AA percentage wide CSV",
        "Dominant AA CSV",
        "Excel summary"
    ],
    "Path": [
        FASTA_ZIP_PATH,
        FASTA_SUMMARY_CSV,
        SITE_PERCENT_LONG_CSV,
        SITE_PERCENT_WIDE_CSV,
        DOMINANT_AA_CSV,
        SITE_PERCENT_EXCEL
    ]
})

display(output_files_df)

display(HTML("<b>Download links:</b>"))

display(HTML("<ul>"))
display(HTML(f"<li>Fasta zip: {FASTA_ZIP_PATH}</li>"))
display(FileLink(FASTA_ZIP_PATH))

display(HTML(f"<li>Excel summary: {SITE_PERCENT_EXCEL}</li>"))
display(FileLink(SITE_PERCENT_EXCEL))

display(HTML(f"<li>AA percentage long CSV: {SITE_PERCENT_LONG_CSV}</li>"))
display(FileLink(SITE_PERCENT_LONG_CSV))

display(HTML(f"<li>AA percentage wide CSV: {SITE_PERCENT_WIDE_CSV}</li>"))
display(FileLink(SITE_PERCENT_WIDE_CSV))

display(HTML(f"<li>Dominant AA CSV: {DOMINANT_AA_CSV}</li>"))
display(FileLink(DOMINANT_AA_CSV))

display(HTML(f"<li>Fasta summary CSV: {FASTA_SUMMARY_CSV}</li>"))
display(FileLink(FASTA_SUMMARY_CSV))
display(HTML("</ul>"))


# ============================================================
# 13. 完成
# ============================================================

display_section("Done")

display(HTML(f"""
<div style="
    background-color:#e2f0d9;
    border:1px solid #70ad47;
    padding:12px;
    border-radius:6px;
">
<b>Finished successfully.</b><br>
Number of sequences: {len(SEQ_DF)}<br>
Number of years: {SEQ_DF[YEAR_COL].nunique()}<br>
Number of positions analyzed: {len(positions_for_percentage)}<br>
Fasta folder: <code>{FASTA_BY_YEAR_DIR}</code><br>
Result folder: <code>{SITE_PERCENT_DIR}</code>
</div>
"""))
```
